# Model Card and Ethical Reflection

This notebook provides a model card and ethical reflection for the COMPAS responsible AI audit.

The purpose of this project is not to recommend the use of recidivism prediction systems. Instead, the purpose is to demonstrate how predictive models can be audited for fairness, subgroup performance, explainability, and responsible AI risks.

## Project Overview

This project examines the COMPAS dataset through the lens of responsible AI.

The project includes:

1. Data understanding  
2. Exploratory data analysis  
3. Baseline model training  
4. Fairness evaluation  
5. Explainability analysis  
6. Model card and ethical reflection  

The central question guiding the project is:

> How can a predictive model be evaluated not only by overall accuracy, but also by subgroup performance, error distribution, explainability, and ethical risk?

## Summary of Main Findings

This project demonstrates why responsible AI evaluation should move beyond aggregate model performance.

The baseline logistic regression model achieved moderate overall performance:

| Metric | Value |
|---|---:|
| Accuracy | 0.675 |
| Precision | 0.662 |
| Recall | 0.567 |
| F1 Score | 0.611 |
| ROC-AUC | 0.722 |

However, the fairness audit showed that aggregate performance does not tell the full story. Error rates differed across demographic groups.

For example, in the test set:

| Group | False Positive Rate | False Negative Rate |
|---|---:|---:|
| African-American | 0.331 | 0.318 |
| Caucasian | 0.166 | 0.591 |
| Hispanic | 0.208 | 0.611 |
| Male | 0.266 | 0.413 |
| Female | 0.145 | 0.534 |

The explainability analysis showed that the model was most strongly influenced by prior offense count and age. Higher prior offense counts increased predicted likelihood of two-year recidivism, while higher age generally lowered predicted likelihood.

The central finding is that model auditing requires multiple layers of evaluation: overall performance, subgroup error rates, explainability, and ethical reflection.

## Intended Use

This project is intended for educational and auditing purposes.

Appropriate uses include:

- Demonstrating responsible AI audit practices.
- Teaching fairness and explainability concepts.
- Comparing aggregate performance with subgroup-level performance.
- Showing how model documentation can support responsible review.

This project should not be used to support actual criminal justice decisions.

## Out-of-Scope Uses

This project should not be used to:

- Recommend detention, sentencing, parole, bail, or supervision decisions.
- Rank individuals by risk in a real criminal justice setting.
- Replace human judgment or legal due process.
- Claim that recidivism can be predicted in a fully objective or neutral way.
- Treat model outputs as causal explanations of human behavior.

## Dataset

The project uses the public COMPAS dataset released by ProPublica.

The dataset contains 7,214 rows and 53 columns. For this project, the target variable is `two_year_recid`, which indicates whether a person was charged with a new offense within two years.

In the full dataset, approximately 45.1% of individuals had `two_year_recid = 1`, while approximately 54.9% had `two_year_recid = 0`.

Key variables used in the baseline model include:

- Age
- Age category
- Juvenile felony count
- Juvenile misdemeanor count
- Juvenile other offense count
- Prior offense count
- Charge degree

Race and sex were not used as model features in the baseline model. However, they were retained for fairness auditing.

This distinction is important because removing protected attributes from a model does not necessarily remove bias. Other features may still reflect structural inequities or act as proxies.

## Model Summary

The project trained baseline classification models to predict two-year recidivism.

Two models were compared:

- Logistic regression
- Random forest

Logistic regression was selected as the primary audit model because it performed better than the random forest model on the main aggregate metrics in this analysis and is relatively interpretable compared with more complex models.

The logistic regression model achieved:

- Accuracy: 0.675
- Precision: 0.662
- Recall: 0.567
- F1 score: 0.611
- ROC-AUC: 0.722

These metrics suggest moderate predictive performance. However, these aggregate values do not show whether the model distributes errors differently across groups. For that reason, the project evaluates false positive and false negative rates by race and sex in the fairness audit.

## Fairness Considerations

The fairness evaluation focused on whether model errors differed across demographic groups.

Particular attention was given to:

- False positive rates by race
- False negative rates by race
- False positive rates by sex
- False negative rates by sex

The subgroup analysis showed that error rates were not evenly distributed.

For race, the African-American subgroup had a higher false positive rate than the Caucasian subgroup in this test set. The African-American false positive rate was approximately 0.331, compared with approximately 0.166 for the Caucasian subgroup. This is significant because false positives represent individuals predicted to recidivate who did not recidivate.

At the same time, the Caucasian and Hispanic subgroups had higher false negative rates than the African-American subgroup. This shows why fairness evaluation is complex in the fact that different groups may experience different types of error.

For sex, the male subgroup had a higher false positive rate than the female subgroup, while the female subgroup had a higher false negative rate than the male subgroup.

These results should be interpreted alongside subgroup sample sizes. The African-American and Caucasian groups had larger sample sizes in the test set, while the Asian and Native American groups had very small sample sizes. Small groups can produce unstable error-rate estimates, so fairness conclusions should avoid overinterpreting metrics for very small subgroups.

## Why Error Type Matters

False positives and false negatives have different ethical consequences.

A false positive means the model predicts recidivism when the person does not recidivate. In a criminal justice context, this type of error could contribute to unnecessary supervision, detention, or harsher treatment.

A false negative means the model predicts no recidivism when the person does recidivate. This type of error may be framed as a public safety concern.

Because these errors carry different consequences, responsible AI auditing should examine both types of error separately instead of relying only on overall accuracy.

## Explainability Considerations

The explainability notebook used logistic regression coefficients and SHAP values to examine which features influenced model predictions.

The global explainability analysis showed that prior offense count and age were the most influential features in the model.

- Higher values of `priors_count` pushed predictions toward a higher likelihood of two-year recidivism.
- Higher age generally pushed predictions toward a lower likelihood of two-year recidivism.
- Juvenile offense counts, age category, and charge degree had smaller effects compared with prior offense count and age.

The local SHAP waterfall example showed how individual features can push a single prediction higher or lower. For that example, age pushed the prediction upward toward class `1`, while prior offense count pushed it downward toward class `0`.

Explainability can help identify which variables are driving predictions, but explainability alone does not make a model fair, valid, or ethically appropriate. A model can be explainable and still produce harmful outcomes.

## Key Limitations

This project has several important limitations:

1. The dataset reflects historical criminal justice data, which may contain structural inequities.
2. The target variable, two-year recidivism, is based on observed charges and may reflect policing, surveillance, and charging patterns rather than behavior alone.
3. The model is a simplified educational baseline and is not optimized for real-world use.
4. Fairness metrics can conflict with one another, so no single metric fully determines whether a model is fair.
5. Subgroup metrics can be unstable for smaller groups.
6. Explainability methods describe model behavior, but they do not establish causality.
7. The project does not include stakeholder input from affected communities.
8. The project does not evaluate legal, procedural, or institutional safeguards that would be necessary in any high-stakes decision-making context.

## Ethical Risk Assessment

This project highlights several ethical risks associated with high-stakes predictive modeling:

- Models may reproduce historical inequities.
- Aggregate accuracy can hide subgroup-level harms.
- Risk scores can appear more objective than they actually are.
- Users may over-rely on model outputs.
- Explanations may create a false sense of transparency.
- Technical fairness metrics may not capture lived experience or institutional harm.

In sensitive domains such as criminal justice, responsible AI requires more than model performance. It requires critical evaluation of whether the system should be built or deployed at all.

## Responsible AI Takeaways

This project demonstrates several responsible AI lessons:

1. Model accuracy is only one part of evaluation.
2. Subgroup error rates are essential for identifying potential harms.
3. Removing protected attributes does not automatically remove bias.
4. Explainability is useful, but it is not the same as fairness.
5. Model cards can help document intended use, limitations, and risks.
6. Some AI systems may be technically possible but ethically inappropriate for deployment.

## Conclusion

This project demonstrates how a responsible AI audit can move beyond traditional model evaluation.

The baseline model achieved moderate aggregate performance, but the fairness audit showed that subgroup error rates varied by race and sex. The explainability analysis showed that the model relied most heavily on prior offense count and age.

These findings reinforce the importance of asking more than whether a model performs well overall. Responsible AI evaluation should also ask:

- Who experiences the model's errors?
- What kinds of errors are most consequential?
- Which features drive predictions?
- What limitations should be documented?
- Should a model like this be used in a high-stakes setting?

The primary contribution of this project is not the predictive model itself, but the audit framework used to evaluate it. In sensitive contexts such as criminal justice, technical performance must be considered alongside fairness, explainability, institutional context, and ethical risk.
